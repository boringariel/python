# 프롬프트 체이닝: ChatGPT가 실수를 줄이고 더 똑똑해지는 비결

---

## 프롬프트 체이닝이란?

**프롬프트 체이닝(Prompt Chaining)**, 또는 프롬프트 연쇄라고도 말하는 개념이 있습니다. 이 것은 챗GPT(ChatGPT) 등의 대규모 자연어 처리 모델(LLM)을 이용할 때, 복잡한 작업을 여러 개의 작은 단계로 나누고, 각 단계의 결과를 다음 단계 입력으로 사용하는 방법입니다. 쉽게 말해, **큰 부탁을 한 번에 하기보다는, 작은 부탁을 여러 번 이어서 하는 것**입니다.  
</p></br></br>

## 프롬프트 체이닝을 쓰는 이유

---

한번에 질문을 해도 되는 문제를 굳이 복잡하게 작은 단계로 나눌 필요는 없습니다. 프롬프트 체이닝은 단계를 나누어서 실행하면 더욱 효과가 좋은 복잡한 문제를 해결해야 할 경우에만 활용해 주세요. 프롬프트 체이닝을 활용할 경우, 아래 요소에서 장점을 보입니다.

* **정확성**: 한 번에 많은 요구사항을 작성하면 ChatGPT가 요구사항 중 몇 가지를 빠뜨리거나 헷갈릴 수 있지만, 단계를 나누면 실수가 줄어듭니다.
* **명확성**: 각 단계에 하나의 목표만 두면, ChatGPT가 더 빠르고 명확하게 답을 줍니다.
* **품질 관리**: 원하는 답이 안 나왔을 때, 프롬프트 전체를 다시 작성할 필요 없이 문제 있는 단계만 다시 시킬 수 있습니다.
</p></br></br>

## 📌 이차방정식이란?

이차방정식(quadratic equation)은 **미지수 (x)의 최고차항이 2차인 방정식**을 말합니다.
일반적인 꼴은 다음과 같습니다.

[
ax^2 + bx + c = 0 \quad (a \neq 0)
]

여기서

* (a, b, c)는 상수
* (a)는 0이 아닌 수
* (x)는 미지수

해를 구하는 방법은 크게 3가지가 있습니다.

1. **인수분해법** → 쉽게 인수분해 되는 경우
2. **완전제곱식 이용**
3. **근의 공식**

[
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
]

---

## 📘 예제 문제

### 문제 1

다음 이차방정식을 풀어라.

[
x^2 - 5x + 6 = 0
]

#### 풀이

[
x^2 - 5x + 6 = (x-2)(x-3) = 0
]
따라서,
[
x = 2 \quad \text{또는} \quad x = 3
]

✅ 정답: (x = 2, 3)

---

### 문제 2

다음 이차방정식을 풀어라.

[
2x^2 - 4x - 6 = 0
]

#### 풀이

먼저 양변을 2로 나눈다.
[
x^2 - 2x - 3 = 0
]
인수분해:
[
(x-3)(x+1) = 0
]
따라서,
[
x = 3 \quad \text{또는} \quad x = -1
]

✅ 정답: (x = 3, -1)

---

### 문제 3

다음 이차방정식을 풀어라.

[
x^2 + 2x + 5 = 0
]

#### 풀이

인수분해가 되지 않으므로 근의 공식을 사용한다.

[
a=1,; b=2,; c=5
]

[
x = \frac{-2 \pm \sqrt{2^2 - 4(1)(5)}}{2(1)}
]

[
x = \frac{-2 \pm \sqrt{4 - 20}}{2}
= \frac{-2 \pm \sqrt{-16}}{2}
]

[
x = \frac{-2 \pm 4i}{2}
= -1 \pm 2i
]

✅ 정답: (x = -1 + 2i, -1 - 2i) (허근)

---

👉 이렇게 해서, 실수해가 나오는 문제(문제 1, 2)와 허근이 나오는 문제(문제 3)까지 같이 다뤄봤습니다.

## ✨ 이차방정식이란?

* **방정식**은 (x) 같은 미지수를 포함하고, 등호(=)가 들어 있는 식이에요.
* 그중에서 **(x^2) 같은 제곱 항(2차항)**이 들어 있는 방정식을 **이차방정식**이라고 해요.

예를 들어,
[
x^2 + 3x + 2 = 0
]
이런 모양을 이차방정식이라고 합니다.

---

## ✨ 이차방정식의 해(답)

* **해**란, 방정식이 참이 되도록 하는 (x)의 값이에요.
* 이차방정식은 보통 해가 **두 개** 나옵니다.
  (특별한 경우에는 하나 또는 허수 해가 나오기도 해요.)

---

## ✨ 푸는 방법

중학교에서는 주로 **인수분해**를 이용해 풀어요.
예)
[
x^2 + 3x + 2 = 0
]

[
(x+1)(x+2) = 0
]

따라서,
[
x = -1, -2
]

---

👉 정리하면, **이차방정식은 미지수 (x)가 제곱으로 들어 있는 방정식이고, 주로 인수분해를 해서 해를 구한다**는 게 핵심이에요.

## 🤖 챗GPT의 세 번째 응답

좋아요 — 각 문제를 **중2 수준**에 맞춰 단계별로 차근차근 풀이해 드릴게요.

# 문제 1

**문제:** 다음 중 이차방정식에 해당하는 것은 무엇인가?
① (2x + 5 = 0)
② (x^2 + 3x - 4 = 0)
③ (3x^3 - 2x = 0)
④ (7 = 0)

**풀이(단계별)**

1. **이차방정식의 정의 확인**: 이차방정식은 (x^2) 항(즉, 최고차항이 2차)이 있는 방정식이다.
2. **각 선택지의 최고차항 확인**

   * ① (2x+5) → 최고차항은 (x^1) (1차) → 이차방정식 아님.
   * ② (x^2+3x-4) → 최고차항은 (x^2) (2차) → 이차방정식 맞음.
   * ③ (3x^3-2x) → 최고차항은 (x^3) (3차) → 이차방정식 아님.
   * ④ (7=0) → 변수가 없음(상수식) → 이차방정식 아님.
3. **정답 도출**: 최고차항이 (x^2)인 ②가 정답이다.

✅ **정답: ②**

---

# 문제 2

**문제:** 다음 이차방정식을 인수분해하여 풀었을 때, 해가 올바르게 짝지어진 것은?
[
x^2 - 5x + 6 = 0
]

**풀이(단계별)**

1. **표준형 확인**: 식이 (x^2 -5x +6 =0)로 이미 표준형(계수 a=1, b=-5, c=6)이다.
2. **인수분해 가능한 두 수 찾기**:

   * 곱이 (c=6)이고 합이 (b=-5)인 두 수를 찾는다.
   * 가능한 쌍: (1,6), (2,3) 및 음수 조합. 합이 -5가 되려면 (-2)와 (-3)이다. ((-2)\times(-3)=6,; -2+(-3)=-5).
3. **인수분해**:
   [
   x^2 -5x +6 = (x-2)(x-3)
   ]
4. **각 인수를 0으로 놓고 해 구하기**:
   [
   x-2=0 \Rightarrow x=2,\quad x-3=0 \Rightarrow x=3
   ]
5. **정답 선택**: (x=2,,3) 이므로 보기 중 ①이 맞다.

✅ **정답: ① ((x=2,3))**

---

# 문제 3

**문제:** 다음 중 해가 **하나만** 존재하는 이차방정식은?
① (x^2 - 4 = 0)
② (x^2 + 2x + 1 = 0)
③ (x^2 + 3x + 2 = 0)
④ (x^2 + 1 = 0)

**풀이(단계별)**

> 중2 수준에서 가장 쉬운 방법은 **판별식(판별식의 의미 설명 포함)** 을 이용하거나, 식이 완전제곱인지 확인하는 것입니다. 판별식 (D=b^2-4ac) 에 따라:
>
> * (D>0)이면 서로 다른 실근 2개,
> * (D=0)이면 중근(같은 해) 1개,
> * (D<0)이면 실근이 없음(허근).

각 선택지별로 판별식이나 인수분해로 확인해 보자.

1. **① (x^2 - 4 = 0)**

   * 계수: (a=1,; b=0,; c=-4)
   * 판별식: (D = 0^2 - 4(1)(-4) = 16 > 0) → 서로 다른 실근 2개
   * (사실 인수분해하면 ((x-2)(x+2)=0) → (x=2,-2)).

2. **② (x^2 + 2x + 1 = 0)**

   * 계수: (a=1,; b=2,; c=1)
   * 판별식: (D = 2^2 -4(1)(1) = 4-4 = 0) → **중근(해가 하나)**
   * (또는 ((x+1)^2=0)로 바로 보고 (x=-1) 하나).

3. **③ (x^2 + 3x + 2 = 0)**

   * 계수: (a=1,; b=3,; c=2)
   * 판별식: (D = 3^2 -4(1)(2) = 9-8 = 1 > 0) → 서로 다른 실근 2개
   * (인수분해 ((x+1)(x+2)=0) → (x=-1,-2)).

4. **④ (x^2 + 1 = 0)**

   * 계수: (a=1,; b=0,; c=1)
   * 판별식: (D = 0^2 -4(1)(1) = -4 < 0) → 실수 해 없음(허근 2개, 복소수 해)

**정리:** 판별식이 0인 경우(②)만 해가 하나이다.

✅ **정답: ②** (해: (x=-1), 중근)
